```python
import pandas as pd

# Load Excel file
file_path = "/mnt/f/Solar-Review-Ppaer/Monthwise-Data-Sheets/21. Dec-2025.xlsx"
df = pd.read_excel(file_path)

# Columns you want to keep
keep_cols = [
    "Time", "Date", "Month", "Year", "Temperature", "Humidity", "Pressure", "GHI"
]

# Keep only those columns (drop rest)
df_filtered = df[keep_cols]

# Show first few rows
print(df_filtered.head())
```

```python
df_filtered
```

```python
import numpy as np

# =============================
# Step 3: Replace '*' with NaN
# =============================
df_filtered.replace('*', np.nan, inplace=True)
```

```python
# 3. Fill NaN with 0
# -----------------------
df_filtered.fillna(0, inplace=True)
```

```python
df_filtered
```

```python
# (Optional) Save to a new Excel file
output_path = "/mnt/f/Solar-Review-Ppaer/Monthwise-Data-Sheets/Data_clean/21. Dec-2025-filtered.xlsx"
df_filtered.to_excel(output_path, index=False)
```
