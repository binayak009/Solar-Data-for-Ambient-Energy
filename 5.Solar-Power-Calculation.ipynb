```python
import pandas as pd

# =============================
file_path = "/mnt/f/Solar-Review-Ppaer/Monthwise-Data-Sheets/Data_clean/10-Minute-Average-Data/20. Dec-2025-filtered-10-min-Avg.xlsx"
df = pd.read_excel(file_path)

# ==============================
# Parameters
# ==============================
ETA = 0.18          # Panel efficiency (18%)
BETA = 0.0045       # Temp coefficient (/°C)
T_REF = 25.0        # Reference temperature
NOCT = 45.0         # Nominal Operating Cell Temperature (°C)
AREA = 1000     # Plant panel area in m²

# ==============================
# Power calculation
# ==============================
# 1) Estimate cell temperature
T_cell = df["Temperature"] + (NOCT - 25.0)/1000.0 * df["GHI"]

# 2) Power per m²
power_density = df["GHI"] * ETA * (1 - BETA * (T_cell - T_REF))
power_density = power_density.clip(lower=0)

# 3) Plant power in MW
df["Power_MW"] = power_density * AREA / 1e3

# ==============================
# Save to same Excel file
# ==============================
with pd.ExcelWriter(file_path, engine="openpyxl", mode="a", if_sheet_exists="overlay") as writer:
    df.to_excel(writer, index=False, sheet_name="Sheet1")
```
